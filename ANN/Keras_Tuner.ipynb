```python
import keras_tuner as kt
```

```python
def build_model(hp):

  model = Sequential()
  model.add(Dense(32,activation='relu',input_dim=8))
  model.add(Dense(16,activation='relu'))
  model.add(Dense(1,activation='sigmoid'))

  optimizer=hp.Choice('Optimerzer',values = ['adam','sgd','rmsprop','adadelta','adamw','adagrad','adamax','adafactor','nadam','ftrl','loin','lamb','muon'])

  model.compile(optimizer=optimizer,loss='binary_crossentropy',metrics=['accuracy'])
  return model
```

```python
tuner = kt.RandomSearch(build_model,objective='val_accuracy',max_trials=10)
```

```python
tuner.search(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,batch_size=32,epochs=100,validation_data=(x_test,y_test),initial_epoch=5)
```

Selecting the Number of Neurons in layer

```python
def build_model_Neuron(hp):

  model2 = Sequential()
  units = hp.Int('units',min_value=8,max_value=64,step=8)

  model2.add(Dense(units=units,activation='relu',input_dim=8))
  model2.add(Dense(1,activation='sigmoid'))

  model2.compile(optimizer='rmsprop',loss='binary_crossentropy',metrics=['accuracy'])
  return model2
```

```python
tuner_Neuron = kt.RandomSearch(
    build_model_Neuron,
    objective='val_accuracy',
    max_trials=10,
)
```

```python
tuner_Neuron.search(
    x_train, y_train,
    epochs=10,
    validation_data=(x_test, y_test)
)
```

```python
tuner_Neuron.get_best_hyperparameters()[0].values
```

Selecting Number of Hidden Layer

```python
def build_model_HiddenLayer(hp):

    model = Sequential()

    # input layer
    model.add(Dense(56, activation='relu', input_shape=(8,), name='inpt_layer'))

    # hidden layers
    for i in range(hp.Int('num_layers', 1, 6)):
        model.add(Dense(
            units=56,
            activation='relu',
            name=f'hidden_{i}'   # ✅ fix here
        ))

    # output layer
    model.add(Dense(1, activation='sigmoid', name='output'))

    model.compile(
        optimizer='rmsprop',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
tuner_hidden = kt.RandomSearch(build_model_HiddenLayer,objective='val_accuracy',max_trials=5,directory='untitled_project',project_name='hiddeen')
```

```python
tuner_hidden.search(x_train,y_train,epochs=10,validation_data=(x_test,y_test))
```

```python
tuner_hidden.get_best_hyperparameters(num_trials=1)[0].values
```

```python
model = tuner_hidden.get_best_models(num_models=1)[0]
```

```python
model.fit(x_train,y_train,batch_size=32,epochs=100,validation_data=(x_test,y_test),initial_epoch=10)
```

Selecting Best Model with Dropout and Early Stopping

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

def build_model_all(hp):

    tf.keras.backend.clear_session()   #  prevents graph issues

    model = Sequential()

    #  Tune number of hidden layers
    num_layers = hp.Int('num_layers', min_value=1, max_value=5)

    for i in range(num_layers):

        #  Tune neurons per layer
        units = hp.Int(f'units_{i}', min_value=8, max_value=128, step=8)

        #  Tune activation function
        activation = hp.Choice(f'activation_{i}', ['relu', 'tanh', 'sigmoid'])

        if i == 0:
            model.add(Dense(
                units=units,
                activation=activation,
                input_shape=(8,),
                name=f'dense_{i}'
            ))
        else:
            model.add(Dense(
                units=units,
                activation=activation,
                name=f'dense_{i}'
            ))

        #  Tune dropout
        dropout_rate = hp.Float(f'dropout_{i}', min_value=0.0, max_value=0.5, step=0.1)
        model.add(Dropout(rate=dropout_rate, name=f'dropout_{i}'))

    #  Output layer
    model.add(Dense(1, activation='sigmoid', name='output'))

    #  Tune optimizer
    optimizer_choice = hp.Choice('optimizer', ['adam', 'rmsprop', 'sgd'])

    #  Tune learning rate
    learning_rate = hp.Choice('learning_rate', [1e-2, 1e-3, 1e-4])

    if optimizer_choice == 'adam':
        optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)
    elif optimizer_choice == 'rmsprop':
        optimizer = tf.keras.optimizers.RMSprop(learning_rate=learning_rate)
    else:
        optimizer = tf.keras.optimizers.SGD(learning_rate=learning_rate)

    model.compile(
        optimizer=optimizer,
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
tuner = kt.RandomSearch(
    build_model_all,
    objective='val_accuracy',
    max_trials=50,   #  increase this
    directory='tuning_dir',
    project_name='full_tuning'
)
```

```python
tuner.search(
    x_train, y_train,
    epochs=30,
    validation_data=(x_test, y_test),
    callbacks=[tf.keras.callbacks.EarlyStopping(patience=5)]
)
```

```python
tuner.get_best_hyperparameters(num_trials=1)[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.summary()
```
