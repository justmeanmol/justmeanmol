```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = pd.read_csv('adm_data (1).xls')
df =df.iloc[:,1:]
df
```

```python
x=df.drop(columns='Chance of Admit ')
y= df['Chance of Admit ']
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
MMS=MinMaxScaler()
x_sclaed=MMS.fit_transform(x)
x_sclaed
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_sclaed,y,test_size=0.2,random_state=30)
```

```python
from tensorflow.keras import Sequential 
from tensorflow.keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(7,activation='relu',input_dim=7))
model.add(Dense(7,activation='relu'))
model.add(Dense(1,activation='linear'))
```

```python
model.summary()
```

```python
model.compile(loss='mean_squared_error',optimizer='Adam',metrics=['r2_score'])
```

```python
model.fit(x_train,y_train,epochs=100,validation_split=0.2,shuffle=False)
```

```python

import tensorflow as tf
print(tf.__version__)
```

```python
y_pred_test = model.predict(x_test)
y_pred_train = model.predict(x_train)
```

```python
r2_score(y_test,y_pred_test)
```

```python
r2_score(y_train,y_pred_train)
```
