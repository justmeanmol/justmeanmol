```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train[0]
```

```python
mean_length = sum(len(item) for item in X_train) / len(X_train)
print(mean_length)
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=200)
X_test = pad_sequences(X_test,padding='post',maxlen=200)
```

```python
X_train[0]
```

```python
model = Sequential()

model.add(SimpleRNN(32,input_shape=(200,1),return_sequences=False))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(X_train,y_train,epochs=50,validation_data=(X_test,y_test))
```
