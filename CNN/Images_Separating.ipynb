```python
import os
import shutil

SOURCE_DIR = "/content/data"
DEST_DIR = "/content/eye_data"

open_dir = os.path.join(DEST_DIR, "open")
closed_dir = os.path.join(DEST_DIR, "closed")

os.makedirs(open_dir, exist_ok=True)
os.makedirs(closed_dir, exist_ok=True)

for root, dirs, files in os.walk(SOURCE_DIR):
    for file in files:
        if file.endswith(".png"):
            label = file.split("_")[4]  # 0=open, 1=closed
            src = os.path.join(root, file)

            if label == "0":
                shutil.copy(src, open_dir)
            elif label == "1":
                shutil.copy(src, closed_dir)

print("Images separated successfully!")
```

```python
print("Open eyes:", len(os.listdir(open_dir)))
print("Closed eyes:", len(os.listdir(closed_dir)))
```
